```python
def xoa_so_nho_hon_x(lst, x):
    lst_new = []
    for i in lst :
        if (i >= x) :
            lst_new.append(i)
    return lst_new

lst = list(map(int, input().split()))
x = int(input())
print(f"list la : {lst} , x = {x}")
xoa_so_nho_hon_x(lst, x)
```

```python
def day_so_0_len(lst):
    so_0 = []
    khac_0 = []
    for i in lst :
        if i == 0 :
            so_0.append(i)
        else :
            khac_0.append(i)
    ans = so_0 + khac_0
    return ans 

lst = list(map(int, input().split()))
print(f"cho list : {lst}")
day_so_0_len(lst)
```

```python
def total_cuc_dai_dp(lst) :
    tong = 0

    for i in range(1, len(lst) - 1):  # bỏ phần tử đầu và cuối
        if lst[i] > lst[i - 1] and lst[i] > lst[i + 1]:
            tong += lst[i]

    print(f"tong la : {tong}")
lst = list(map(int, input().split()))
print(f"cho list : {lst}")
total_cuc_dai_dp(lst)
```
