```python
def is_armstrong(x):
    tung_so = str(x)
    mu = len(tung_so) #len chi dung duoc khi di voi str or lst or tuple

    total = 0
    for i in tung_so :
        total += pow(int(i), mu)

    return total == x

n = int(input())
print(f"{n} la so armstrong : {is_armstrong(n)}")
```

```python
def total_uoc(n):
    total = 0
    uoc = []
    for i in range(1,n):
        if n % i == 0 :
            uoc.append(i)
    for i in uoc :
        total += i
    return total

n = int(input())
if total_uoc(n) == n :
    print(f"{n} la so ky dieu")
else :
    print(f"{n} khong phai so ky dieu")
```

```python
def quadratic_roots_count(a,b,c):
    delta = pow(b,2) - 4*a*c
    can_delta = pow(delta, 0.5)
    if delta < 0 :
        print(f"phuong trinh vo nghiem")
    elif delta == 0 :
        x = -b / 2*a
        print(f"phuong trinh co nghiem kep la : {x}")
    elif delta > 0 :
        x1 = (-b + can_delta ) / (2*a)
        x2 = (-b - can_delta ) / (2*a)
        print(f"phuong trinh co 2 nghiem la {x1} va {x2}")

a,b,c = map(float, input().split())
print(f"nhap 3 so a,b,c : {a} {b} {c}")
quadratic_roots_count(a,b,c)
```

```python
def area(A, B, C): #tinh dien tich tam giac bang tich co huong
    return abs((B[0] - A[0]) * (C[1] - A[1]) - (C[0] - A[0]) * (B[1] - A[1]))

def point_in_triangle(A, B, C, P): # xet xem dien tich co dung khong
    S = area(A, B, C)
    S1 = area(P, B, C)
    S2 = area(A, P, C)
    S3 = area(A, B, P)
    return S == S1 + S2 + S3 #true or false

def point_in_quadrilateral(quad, p):
    A, B, C, D = quad
    # chia thanh 2 tam giac: ABC và ACD
    return point_in_triangle(A, B, C, p) or point_in_triangle(A, C, D, p)

p = tuple(map(int, input("Nhập p (x y): ").split()))

quad = []
for i in range(4):
    x, y = map(int, input(f"Nhập điểm {i+1} (x y): ").split())
    quad.append((x, y))

print(point_in_quadrilateral(quad, p))
```

```python
import math
def goc(A,B,C): # tinh theo cos x = tich vo huong / tich do dai
    (x1,y1) = A
    (x2,y2) = B
    (x3,y3) = C
    BA = (x1-x2,y1-y2)
    BC = (x3-x2,y3-y2)
    tich_vo_huong = BA[0]*BC[0]+BA[1]*BC[1]
    abs_BA = (BA[0]**2+BA[1]**2)**0.5
    abs_BC = (BC[0]**2+BC[1]**2)**0.5
    cos_x = tich_vo_huong / (abs_BA * abs_BC)
    x_rad = math.acos(cos_x)
    x_do = math.degrees(x_rad)
    return x_do

A = tuple(map(float, input("Nhập A(x y): ").split()))
B = tuple(map(float, input("Nhập B(x y): ").split()))
C = tuple(map(float, input("Nhập C(x y): ").split()))

print("Góc ABC =", goc(A,B,C), "độ")
```

```python
def giai_thua(x):
    if x == 1 :
        return 1
    else :
        return x * giai_thua(x-1)
    
def C(n,k):
    x = giai_thua(n) / (giai_thua(k)*giai_thua(n-k))
    return x

while (True):
    n , k = map(int, input().split()) #nhap n > k
    if n > k :
        break
print(C(n,k))


```

```python
def samrt_taxi_fare(car_type,start_hour,km,wait_min):
    fare = 0
    if car_type == "4cho":
        base_fare = 10000
        price_per_km = 12000
    elif car_type == "7cho":
        base_fare = 12000
        price_per_km = 15000
    else:
        return "Loại xe không hợp lệ"

    fare = base_fare + km * price_per_km #tien quang duong di 

    fare += wait_min * 1000 # phi cho

    if start_hour >= 22 or start_hour < 5: #phu phi ban dem tang 20%
        fare *= 1.2

    return fare

car_type = input()
start_hour,km,wait_min = map(int, input().split())
print(f"ten loai xe (4cho) or (7cho) : {car_type}")
print(f"gio di : {start_hour}")
print(f"so km di : {km}")
print(f"so phut doi : {wait_min}")
print(f"tong so tien phai tra la : {samrt_taxi_fare(car_type,start_hour,km,wait_min)}")

```

```python
def sum_1_to_n(n):
    lst = []
    total = 0
    for i in range (1 , n+1):
        if len(str(i)) == len(set(str(i))): # doan nay la loc chu so doi 1 trung nhau
            lst.append(i)
    for i in lst :
        total += i
    return total

n = int(input())
if n <= 0 :
    print(f"0")
else :
    print(f"tong cac so tu 1 den {n} co chu so doi 1 khac nhau la : {sum_1_to_n(n)}")
```
