```python
s = input()

ans = ""
i = 0

# bo khoang trang dau 
while i < len(s) and i == " ":
    i += 1 

# dung de xem co khoang trang phia truoc hay chua va co phai tu moi khong 
space = False
new_word = True

while i < len(s):
    #neu la khoang trang
    if s[i] == " ":
        #chi them 1 dau cach
        if not space :
            ans += " "
            space = True

        new_word = True # dung de xac dinh day la sang tu moi
    
    else :
        #viet hoa chu cai dau cua moi tu moi
        if new_word and "a" <= s[i] <= "z":
            ans += chr(ord(s[i]) - 32)

        #viet thuong chu cai giua 
        elif not new_word and "A" <= s[i] <= "Z":
            ans += chr(ord(s[i]) + 32)
        
        #th nay la dung roi
        else :
            ans += s[i]

        space = False
        new_word = False

    i += 1 

#bo khoang trang cuoi
if len(ans) > 0 and ans[-1] == " ":
    ans = ans[:-1]

print(ans)
```

```python
def partition(arr, left, right):
    pivot = arr[right]
    i = left

    for j in range(left, right):
        if arr[j] > pivot:   # lớn hơn để tìm số lớn thứ k
            arr[i], arr[j] = arr[j], arr[i]
            i += 1

    arr[i], arr[right] = arr[right], arr[i]
    return i


def quick_select(arr, left, right, k):
    if left <= right:
        pos = partition(arr, left, right)

        if pos == k:
            return arr[pos]

        elif pos > k:
            return quick_select(arr, left, pos - 1, k)

        else:
            return quick_select(arr, pos + 1, right, k)


arr = list(map(int, input().split()))
k = int(input())

result = quick_select(arr, 0, len(arr) - 1, k - 1)

print(result)
```

```python
n = str(input())
k = int(input())

#so chu so can xoa
remove_count = len(n) - k

#them vao chuoi
stack = []

#xoa phan tu ben trai neu phan tu ben phai lon hon 
for ch in n :
    while stack and remove_count > 0 and stack[-1] < ch: #stack con ptu , van con phan tu can xoa , phan tu cuoi cua stack nho hon ch thi thuc thi
        stack.pop()
        remove_count -= 1 
    stack.append(ch)

# xoa trong truong hop day giam 
while remove_count > 0:
    stack.pop()
    remove_count -= 1

print("".join(stack))
```

```python
s = input()
print(s)
special = "!@#$"

has_lower = False
has_upper = False
has_digit = False
has_special = False

if len(s) >= 8:
    for ch in s :
        if ch.islower():
            has_lower = True
        if ch.isupper():
            has_upper = True
        if ch.isdigit():
            has_digit = True
        if ch in special:
            has_special = True

    if has_special and has_digit and has_lower and has_upper :
        print(f"Strong password")
    else :
        print(f"Weak password")
else :
    print(f"Weak password")
```
