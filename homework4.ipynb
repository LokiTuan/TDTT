W4A20 

```python
#xet n duong
while (True):
    n = int(input())
    if (n > 0) :
        break
#xet xem 2^k
temp = n 
while temp > 1  and temp % 2 == 0:
    temp //= 2 
if temp == 1 :
    print(f"{n} là số dạng 2^k")
else:
    print(f"{n} không phải là số dạng 2^k")
```

```python
#xet n duong
while (True):
    n = int(input())
    if (n > 0) :
        break 
#tong cac chu so
tong = 0 
temp = n 
while (temp>0):
    lay_chu_so = temp % 10
    tong += lay_chu_so
    temp //= 10

print(f"tong cac chu so la {tong}")
```
