```python
def merge_list(a,b):
    result = []
    i = 0
    j = 0

    while i < len(a) and j < len(b): # dung de them ptu vao thu cong
        if a[i] < b[j]:
            result.append(a[i])
            i += 1 
        else :
            result.append(b[j])
            j += 1

    while i < len(a): # chay khi a con ma b het
        result.append(a[i])
        i += 1
    
    while j < len(b): # chay khi a het ma b con
        result.append(b[j])
        j += 1 
    
    return result

l1 = list(map(int, input().split()))
l2 = list(map(int, input().split()))
print(f"{l1}\n{l2}")
print(f"merge cua 2 list la : {merge_list(l1, l2)}")
```

```python
def tim_so_lon_k(matrix, k):
    ds = []
    for row in matrix:
        for x in row:
            ds.append(x)
    ds.sort(reverse=True)
    return ds[k-1] # vi no chay tu 0

n , m = list(map(int, input().split()))
k = int(input())
matrix = []

#tao ma tran
for i in range(n):
    row = list(map(int, input().split()))
    matrix.append(row)

print(f"cac hang cua ma tran la : {matrix}")
print(f"k la {k}")
print(f"so lon thu k la : {tim_so_lon_k(matrix, k)}")
```

```python
def so_nho_nhat(n):
    s = list(str(n))
    min_num = n

    #hoan doi tung so 1 
    for i in range(len(s)):
        for j in range(i+1 , len(s)):
            temp = s.copy()
            temp[i] , temp[j] = temp[j] , temp[i]

            if temp[0] == "0":
                continue

            new_num = int("".join(temp))

            if new_num < min_num:
                min_num = new_num
    return min_num

n = int(input())
print(f"cho n : {n}")
print(f"so nho nhat la : {so_nho_nhat(n)}")
```

```python
def sum_to_n(n):
    if n == 1 :
        return 1 
    else :
        return n + sum_to_n(n-1)
    
n = int(input())
print(f"n la : {n}, tong la : {sum_to_n(n)}")
```

```python
def power(base,exp):
    if exp == 0 :
        return 1 
    else :
        return base * power(base, exp-1)

n , k = map(int, input().split())
print(f"{n} mu {k} la bang : {power(n,k)}")
```

```python
def so_xh_nhieu_x(lst, x):
    ans = []
    for so in lst :
        if lst.count(so) > x and so not in ans :
            ans.append(so)
        
    if len(ans) == 0 :
        return f"INVALID"
    else :
        return ans 

lst = list(map(int, input().split()))
x = int(input())

print(f"ta co list la : {lst}, x la {x}")
print(f"so xuat hien nhieu hon x lan la : {so_xh_nhieu_x(lst,x)}")
```

```python
def day_tang_max(lst):
    arr = []
    lst_new = [lst[0]]
    for i in range(1, len(lst)):
        if lst[i] > lst[i-1]:
            lst_new.append(lst[i])
        else :
            if len(lst_new) > len(arr):
                arr = lst_new
            lst_new = [lst[i]]

    if len(lst_new) > len(arr):
        arr = lst_new    
    return arr

def la_mot_day_tang_or_giam_nn(lst):
    day_tang = []
    day_giam = []
    for i in range(1, len(lst)):
        if lst[i] > lst[i-1]:
            day_tang.append(lst[i])
        if lst[i] < lst[i-1]:
            day_giam.append(lst[i])
    if len(day_tang) == len(lst)-1 :
        return f"day la day tang nghiem ngat"
    elif len(day_giam) == len(lst)-1:
        return f"day la day giam nghiem ngat"
    else :
        return f"day khong phai la day tang hay giam nghiem ngat \nday con tang dai nhat la : {day_tang_max(lst)}"


lst = list(map(int , input().split()))
print(f"lst da cho la : {lst}")
print(la_mot_day_tang_or_giam_nn(lst))
```
