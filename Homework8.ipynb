```python
class shape:
    def area(self):
        return 0

class Rectangle(shape):
    def __init__(self, w , h):
        self.w = w
        self.h = h
    
    def area(self):
        return self.w * self.h 
    
class Circle(shape):
    def __init__(self, r):
        self.r = r
    
    def area(self):
        return 3.14 * pow(self.r, 2)
    
shape_type = input()

if shape_type == "RECTANGLE" :
    w , h = map(float, input().split())
    obj = Rectangle(w,h)

if shape_type == "CIRCLE" :
    r = float(input())
    obj = Circle(r)

print(f"{obj.area():.2f}")

```

```python
class Vehicle :
    def __init__(self , brand , speed):
        self.brand = brand
        self.speed = speed
    
    def info():
        return f"{self.brand}-{self.speed}"

class Car(Vehicle):
    def __init__(self, brand, speed, seats):
        super().__init__(brand, speed)
        self.seats = seats

    def info(self):
        return f"{self.brand}-{self.speed}-{self.seats}"

obj_type = input()

if obj_type == "Vehicle" :
    brand , speed = input().split()
    obj = Vehicle(brand, speed)

if obj_type == "Car" :
    brand , speed , seats = input().split()
    obj = Car(brand , speed , seats) 

print(obj.info())
```

```python
class Polygon :
    def perimeter(self):
        return 0 
    
class Triangle(Polygon):
    def __init__(self, a , b , c):
        self.a = a
        self.b = b
        self.c = c 
    
    def perimeter(self):
        return self.a + self.b + self.c
    
a , b , c = map(int , input().split())
chu_vi = Triangle(a,b,c)
print(chu_vi.perimeter())
```

```python
class Person :
    def __init__(self , name , age):
        self.name = name 
        self.age = age 

    def describe(self):
        return f"{self.name} - {self.age}"
    
class Student(Person):
    def __init__(self, name, age , school):
        super().__init__(name, age)
        self.school = school

    def describe(self):
        return f"{self.name} - {self.age} - {self.school}"
    
obj_type = input()

if obj_type == "Person":
    name , age = input().split()
    info = Person(name , age)

if obj_type == "Student" :
    name , age , school = input().split()
    info = Student(name , age , school)

print(info.describe())
        
```

```python
class Account :
    def __init__(self, balance):
        self.balance = balance
    
    def deposit(self , x):
        self.balance += x 
    
    def get_balance(self):
        return self.balance
    
class SavingAccount(Account):
    def __init__(self, balance , rate):
        super().__init__(balance)
        self.rate = rate 
    
    def add_interest(self):
        self.balance += self.balance * self.rate

balance , rate = map(float, input().split())
acc = SavingAccount(balance, rate)
acc.add_interest()
print(f"{acc.get_balance():.2f}")
```

```python
class Printer :
    def __init__(self):
        pass

    def print_value(self, x):
        print(x)

class BinaryPrinter(Printer):
    def __init__(self):
        super().__init__()
    
    def print_value(self, x):
        print(bin(x)[2:])

while(True):
    n = int(input())
    if n > 0 :
        break 
obj_bin = BinaryPrinter()
obj_bin.print_value(n)
```

```python
class Matrix2x2:
    def __init__(self, a ,b ,c ,d):
        self.a = a
        self.b = b
        self.c = c
        self.d = d 

    def trace(self):
        return self.a + self.d
    
    def determinant(self):
        return self.a * self.d - self.b * self.c

a , b , c , d = map(int, input().split())
matrix = Matrix2x2(a,b,c,d)
print(f"{matrix.trace()} , {matrix.determinant()}")        
```

```python
class Product:
    def __init__(self, name , price):
        self.name = name
        self.price = price

    def final_price(self):
        return self.price
    
class DiscountProduct(Product):
    def __init__(self, name, price, percent):
        super().__init__(name, price)
        self.percent = percent

    def final_price(self):
        price_last = int(price) * (100 - int(percent)) / 100
        return f"{price_last:.2f}"
    
name , price , percent = input().split()
tien = DiscountProduct(name,price,percent)
print(tien.final_price())

        
```

```python
class Counter:
    def __init__(self):
        self.count = 0

    def inc(self):
        self.count += 1

    def dec(self):
        self.count -= 1

    def value(self):
        return self.count


n = int(input())
operations = input().split()

counter = Counter()

for op in operations:
    if op == "INC":
        counter.inc()
    elif op == "DEC":
        counter.dec()

print(counter.value())
```

```python
class Time:
    def __init__(self,h,m,s):
        self.h = h 
        self.m = m
        self.s = s 

    def to_second(self):
        return self.h * 3600 + self.m * 60 + self.s
    
h,m,s = map(float, input().split())
gio = Time(h,m,s)
print(gio.to_second())
        
```

```python
class Stack :
    def __init__(self):
        self.data = []
    
    def push(self,x):
        self.data.append(x)
    
    def pop(self):
        self.data.pop()

    def top(self):
        return self.data[-1]
    
n = int(input())
stack = Stack()
lst = []
for _ in range (n):
    comand = input().split()
    lst.append(comand)
    if comand[0] == "PUSH":
        stack.push(int(comand[1]))

    elif comand[0] == "POP":
        stack.pop()

    elif comand[0] == "TOP":
        print(stack.top())
print(lst)
```
