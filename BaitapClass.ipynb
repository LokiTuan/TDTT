```python
class Cylinder:
    def __init__(self,r,h):
        self.r = r
        self.h = h
    
    def getarea(self):
        return 2 * 3.14 * self.r * self.h + 2 * 3.14 * pow(self.r,2)
    
    def getvolume(self):
        return 3.14 * pow(self.r,2) * h
    
r , h = map(float, input().split())
cy1 = Cylinder(r, h)
print(f"dien tich hinh tru : {cy1.getarea()}")
print(f"the tich hinh tru : {cy1.getvolume()}")


```

```python
class Date:
    def __init__(self,d ,m ,y):
        self.d = d
        self.m = m 
        self.y = y
    
    def d_m_y_hople(self):
        maxDay = -1
        match self.m:
            case 1|3|5|7|8|10|12:
                maxDay = 31
            case 4|6|9|11:
                maxDay = 30
            case 2:
                if (self.y%4==0 and self.y%100!=0) or (self.y%400==0):
                    maxDay = 29
                else :
                    maxDay = 28
        return (self.m > 0 and (1 <= self.d <= maxDay))
    
    def Getnextday(self):
        nextday = Date(self.d + 1, self.m , self.y)
        if nextday.d_m_y_hople():
            return nextday
        
        nextday.d = 1
        nextday.m += 1 
        if (nextday.d_m_y_hople()):
            return nextday
        else :
            nextday.d = 1 
            nextday.m =1 
            nextday.y += 1
            if (nextday.d_m_y_hople()):
                return nextday
    
    def __str__(self):
        # Định dạng hiển thị ngày/tháng/năm khi in
        return f"{self.d}/{self.m}/{self.y}"

        

d , m , y = map(int, input().split("/"))
day = Date(d,m,y)
print(f"day la ngay hop le : {day.d_m_y_hople()}")
print(f"ngay ke tiep la : {day.Getnextday()}")
```

```python
class Toado:
    def __init__(self,x,y):
        self.x = x
        self.y = y
    
    def checkO(self):
        return self.x == 0 and self.y == 0
    
    def checkX(self):
        return self.x != 0 and self.y == 0
    
    def checkY(self):
        return self.x == 0 and self.y != 0
    
    def distance(self):
        bp = pow(self.x,2) + pow(self.y,2)
        return pow(bp, 1/2)
    
x,y = map(int, input().split())
M = Toado(x,y)
print(f"ta co toa do diem M la : M({x},{y})")
if M.checkO():
    print(f"M nam o goc toa do")
if M.checkX():
    print(f"M nam tren truc hoanh voi x = {x}")
if M.checkY():
    print(f"M nam tren truc tung voi y = {y}")
print(f"khoang cach tu M toi goc toa do la : {M.distance()}")

        
```

```python
class Caculator:
    def __init__(self,a,b):
        self.a = a
        self.b = b

    def add(self):
        return a + b
    
    def subtract(self):
        return a - b
    
    def product(self):
        return a * b
    
    def divide(self):
        if self.b == 0:
            return "Không thể chia cho 0"
        return a // b
    
    def power(self):
        return pow(a,b)
    
    def mod(self):
        if self.b == 0:
            return "Không thể chia cho 0"
        return a % b
    
    def set_numbers(self,a,b):
        self.a = a
        self.b = b

a,b = map(int , input().split())
calc = Caculator(a,b)
while True:
    print("\n===== MENU =====")
    print("1. Cộng")
    print("2. Trừ")
    print("3. Nhân")
    print("4. Chia")
    print("5. Lũy thừa")
    print("6. Chia lấy dư")
    print("7. Nhập 2 số mới")

    choice = input("Chọn phép toán: ")

    if choice == "1":
        print("Kết quả:", calc.add())

    elif choice == "2":
        print("Kết quả:", calc.subtract())

    elif choice == "3":
        print("Kết quả:", calc.product())

    elif choice == "4":
        print("Kết quả:", calc.divide())

    elif choice == "5":
        print("Kết quả:", calc.power())

    elif choice == "6":
        print("Kết quả:", calc.mod())

    elif choice == "7":
        a = int(input("Nhập số thứ nhất mới: "))
        b = int(input("Nhập số thứ hai mới: "))
        calc.set_numbers(a, b)
        print("Đã cập nhật số mới!")

    else:
        print("Lựa chọn không hợp lệ!")

    stop = input("Bạn có muốn thoát? ")
    if stop.lower() in ["yes", "y"]:
        break
    

        
```

```python
class Rectangle:
    def __init__(self,w,h):
        self.w = w 
        self.h = h

    def area(self):
        return self.w*self.h
    
    def perimeter(self):
        return 2 * (self.w + self.h)
    
    def scale(self, k):
        self.w = w*k
        self.h = h*k
        return f"do dai moi la {self.w}x{self.h}"

w,h,k = map(int, input().split())
rec = Rectangle(w,h)
print(f"do dai ban dau la {w}x{h}")
print(rec.scale(k))
print(f"dien tich moi la : {rec.area()}")
print(f"chu vi moi la : {rec.perimeter()}")



        
```

```python
class Fraction:
    def __init__(self,num,den):
        a,b = abs(num),abs(den)
        while b:
            a,b = b , a%b
        gcd = a
        self.num = num // gcd
        self.den = den // gcd 
        if self.den < 0 :
            self.num = -self.num
            self.den = -self.den
    
    def add(self,others):
        return Fraction(self.num * others.den + others.num * self.den, self.den * others.den)
    
    def sub(self,others):
        return Fraction(self.num * others.den - others.num * self.den, self.den * others.den)
    
    def mul(self,others):
        return Fraction(self.num * others.num , self.den * others.den)
    
    def div(self,others):
        return Fraction(self.num * others.den , self.den * others.num)
    
    def __str__(self):
        return f"{self.num}/{self.den}"
    
arr = list(input().split())
ps1 = Fraction(int(arr[0]),int(arr[1]))
ps2 = Fraction(int(arr[3]),int(arr[4]))
op = arr[2]

if op == "+":
    print(ps1.add(ps2))
if op == "-":
    print(ps1.sub(ps2))
if op == "*":
    print(ps1.mul(ps2))
if op == "/":
    print(ps1.div(ps2))
```

```python
class BankAccount:
    def __init__(self,owner,balance):
        self.owner = owner
        self.balance = balance

    def deposit(self,x):
        self.balance += x
        return self.balance
    
    def withdraw(self,x):
        if self.balance >= x :
            self.balance -= x
        return self.balance
    
    def __str__(self):
        return f"{self.balance}"
    
owner , balance = input().split()
ba = int(balance)
obj = BankAccount(owner,ba)
n = int(input())
for i in range(n):
    cmd , tien = input().split()
    x = int(tien)
    if cmd == "DEPOSIT":
        obj.deposit(x)
    if cmd == "WITHDRAW":
        obj.withdraw(x)
print(f"{obj.__str__():.2f}")
        
```

```python
class Student:
    def __init__(self,name):
        self.name = name 
        self.scores = []

    def add_score(self,subject,score):
        self.scores.append(score)

    def avg(self):
        total = 0
        for i in range(len(self.scores)):
            total += self.scores[i]
        return total / len(self.scores)
    
    def rank(self):
        average = self.avg()
        if average >= 8:
            return "Excellent"
        elif average >= 6.5:
            return "Good"
        elif average >= 5:
            return "Average"
        else:
            return "Poor"

name = input()
hs = Student(name)
n = int(input())
for i in range(n):
    subject , diem = input().split()
    scores = int(diem)
    hs.add_score(subject,scores)
print(f"{name} {hs.avg()}")
print(f"{hs.rank()}")

        
```

```python
class Book:
    def __init__(self,title,author,year):
        self.title = title
        self.author = author
        self.year = year

class Library():
    def __init__(self):
        self.book = []
    
    def add(self,book):
        self.book.append(book)

    def count_by_author(self,name):
        count = 0
        for book in self.book:
            if book.author == name :
                count += 1 
        return count
    
    def find_by_year(self,y):
        count = 0
        for book in self.book:
            if book.year == y:
                count += 1
        return count    
n = int(input())
library = Library()

for _ in range(n):
    cmd, data = input().split()

    if cmd == "ADD":
        title, author, year = data.split(";")
        library.add(Book(title, author, int(year)))

    elif cmd == "COUNT":
        print(library.count_by_author(data))

    elif cmd == "COUNTYEAR":
        print(library.find_by_year(int(data)))
```
